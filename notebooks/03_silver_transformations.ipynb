```python
orders_df = spark.table("bronze_orders")
restaurants_df = spark.table("bronze_restaurants")
riders_df = spark.table("bronze_riders")
refunds_df = spark.table("bronze_refunds")
```

```python
display(orders_df)
```

```python
orders_df.printSchema()
```

```python
from pyspark.sql.functions import col, trim, upper, initcap

orders_std = orders_df \
    .withColumn("city", initcap(trim(col("city")))) \
    .withColumn("delivery_zone", upper(trim(col("delivery_zone")))) \
    .withColumn("final_status", upper(trim(col("final_status")))) \
    .withColumn("payment_method", upper(trim(col("payment_method")))) \
    .withColumn("payment_status", upper(trim(col("payment_status")))) \
    .withColumn("required_vehicle_category", upper(trim(col("required_vehicle_category")))) \
    .withColumn("time_band", upper(trim(col("time_band")))) \
    .withColumn("source_system", upper(trim(col("source_system"))))
```

```python
orders_std.printSchema()
```

```python
from pyspark.sql.functions import unix_timestamp

orders_calc = orders_std.withColumn(
    "delivery_time_minutes",
    (
        unix_timestamp(col("delivered_ts")) -
        unix_timestamp(col("picked_up_ts"))
    ) / 60
)
```

```python
orders_calc = orders_calc.withColumn(
    "gross_order_value_inr",
    col("net_order_value_inr") + col("discount_amount_inr")
)
```

```python
orders_calc = orders_calc.withColumn(
    "total_charges",
    col("delivery_fee_inr") + col("tax_amount_inr")
)
```

```python
orders_calc.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("silver_orders_candidate")
```

```python
restaurants_df.printSchema()
```

```python
from pyspark.sql.functions import col, trim, upper, initcap

restaurants_std = restaurants_df \
    .withColumn("city", initcap(trim(col("city")))) \
    .withColumn("cuisine_primary", upper(trim(col("cuisine_primary")))) \
    .withColumn("delivery_zone", upper(trim(col("delivery_zone")))) \
    .withColumn("restaurant_name", initcap(trim(col("restaurant_name")))) \
    .withColumn("service_status", upper(trim(col("service_status")))) \
    .withColumn("zone_group", upper(trim(col("zone_group"))))
```

```python
from pyspark.sql.functions import to_date,to_timestamp

restaurants_typed = restaurants_std \
    .withColumn("opened_date",
                to_date(col("opened_date"))) \
    .withColumn("record_updated_at",
                to_timestamp(col("record_updated_at")))
```

```python
from pyspark.sql.functions import current_date,datediff

restaurants_calc = restaurants_typed.withColumn(
    "restaurant_age_days",
    datediff(current_date(),col("opened_date"))
)
```

```python
restaurants_calc = restaurants_calc.withColumn(
    "minimum_order_with_tax",
    col("min_order_value_inr") * 1.05
)
```

```python
restaurants_calc.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("silver_restaurants_candidate")
```

```python
riders_df.printSchema()
```

```python
from pyspark.sql.functions import col, trim, upper, initcap

riders_std = riders_df \
    .withColumn("home_city", initcap(trim(col("home_city")))) \
    .withColumn("service_zone", upper(trim(col("service_zone")))) \
    .withColumn("zone_group", upper(trim(col("zone_group")))) \
    .withColumn("vehicle_type", upper(trim(col("vehicle_type")))) \
    .withColumn("service_status", upper(trim(col("service_status")))) \
    .withColumn("shift_type", upper(trim(col("shift_type"))))
```

```python
from pyspark.sql.functions import col, to_date, to_timestamp

riders_typed = riders_std \
    .withColumn(
        "max_service_radius_km",
        col("max_service_radius_km").cast("double")
    ) \
    .withColumn(
        "onboarding_date",
        to_date(col("onboarding_date"))
    ) \
    .withColumn(
        "record_updated_at",
        to_timestamp(col("record_updated_at"))
    )
```

```python
from pyspark.sql.functions import current_date, datediff

riders_calc = riders_typed.withColumn(
    "experience_days",
    datediff(current_date(), col("onboarding_date"))
)
```

```python
riders_calc.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("silver_riders_candidate")
```

```python
%sql
SELECT COUNT(*) FROM silver_riders_candidate;
```

```python
refunds_df.printSchema()
```

```python
from pyspark.sql.functions import col, trim, upper

refunds_std = refunds_df \
    .withColumn("refund_reason", upper(trim(col("refund_reason")))) \
    .withColumn("refund_status", upper(trim(col("refund_status")))) \
    .withColumn("refund_channel", upper(trim(col("refund_channel")))) \
    .withColumn("source_system", upper(trim(col("source_system"))))
```

```python
from pyspark.sql.functions import to_timestamp

refunds_typed = refunds_std \
    .withColumn(
        "refund_requested_ts",
        to_timestamp(col("refund_requested_ts"))
    ) \
    .withColumn(
        "refund_approved_ts",
        to_timestamp(col("refund_approved_ts"))
    ) \
    .withColumn(
        "refund_amount_inr",
        col("refund_amount_inr").cast("double")
    )
```

```python
from pyspark.sql.functions import unix_timestamp

refunds_calc = refunds_typed.withColumn(
    "refund_processing_minutes",
    (
        unix_timestamp(col("refund_approved_ts")) -
        unix_timestamp(col("refund_requested_ts"))
    ) / 60
)
```

```python
refunds_calc.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("silver_refunds_candidate")
```
