```python
%sql
DESCRIBE trusted_silver_orders;
```

```python
%sql
SELECT *
FROM trusted_silver_orders
LIMIT 10;
```

```python
%sql
SELECT
    restaurant_id,
    order_placed_ts,
    order_id
FROM trusted_silver_orders
LIMIT 10;
```

```python
%sql
SELECT
    COUNT(*) AS total_orders,
    COUNT(order_placed_ts) AS eligible_orders,
    COUNT(*) - COUNT(order_placed_ts) AS outside_scope_orders
FROM trusted_silver_orders;
```

```python
%sql
SELECT
    restaurant_id,
    COUNT(*) AS row_count
FROM trusted_silver_restaurants
GROUP BY restaurant_id
HAVING COUNT(*) > 1;
```

```python
%sql
SELECT
    COUNT(*) AS orders_before_join,
    COUNT(r.restaurant_id) AS matched_restaurants,
    COUNT(*) - COUNT(r.restaurant_id) AS unmatched_restaurants
FROM trusted_silver_orders o
LEFT JOIN trusted_silver_restaurants r
    ON o.restaurant_id = r.restaurant_id;
```

```python
%sql
CREATE OR REPLACE TABLE gold_daily_restaurant_orders
USING DELTA
AS
SELECT
    o.restaurant_id,
    DATE(o.order_placed_ts) AS order_date,
    COUNT(*) AS daily_order_count
FROM trusted_silver_orders o
GROUP BY
    o.restaurant_id,
    DATE(o.order_placed_ts);
```

```python
%sql
SELECT *
FROM gold_daily_restaurant_orders
LIMIT 10;
```

```python
%sql
SELECT
    restaurant_id,
    order_date,
    COUNT(*) AS row_count
FROM gold_daily_restaurant_orders
GROUP BY
    restaurant_id,
    order_date
HAVING COUNT(*) > 1;
```

```python
%sql
SELECT
    COUNT(*) AS total_gold_rows,
    SUM(CASE WHEN daily_order_count <= 0 THEN 1 ELSE 0 END) AS invalid_counts,
    MIN(daily_order_count) AS minimum_order_count,
    MAX(daily_order_count) AS maximum_order_count
FROM gold_daily_restaurant_orders;
```

```python
%sql
SELECT
    (SELECT COUNT(*)
     FROM trusted_silver_orders) AS silver_order_count,

    (SELECT SUM(daily_order_count)
     FROM gold_daily_restaurant_orders) AS gold_order_count;
```

```python
%sql
SELECT
    (SELECT COUNT(*) FROM trusted_silver_orders) AS silver_orders,
    (SELECT SUM(daily_order_count)
     FROM gold_daily_restaurant_orders) AS gold_orders,
    (SELECT COUNT(*) FROM trusted_silver_orders)
      - (SELECT SUM(daily_order_count)
         FROM gold_daily_restaurant_orders) AS difference;
```

```python
%sql
CREATE OR REPLACE TABLE gold_daily_restaurant_orders
USING DELTA
AS
SELECT
    o.restaurant_id,
    DATE(o.order_placed_ts) AS order_date,
    COUNT(*) AS daily_order_count
FROM trusted_silver_orders o
GROUP BY
    o.restaurant_id,
    DATE(o.order_placed_ts);
```

```python
%sql
SELECT
    COUNT(*) AS total_gold_rows,
    SUM(daily_order_count) AS total_orders
FROM gold_daily_restaurant_orders;
```

```python
%sql

SELECT
    COUNT(*) AS gold_rows,
    MIN(daily_order_count) AS min_daily_orders,
    MAX(daily_order_count) AS max_daily_orders,
    SUM(daily_order_count) AS total_orders
FROM gold_daily_restaurant_orders;
```

```python
%sql

SELECT
    (SELECT COUNT(*) FROM trusted_silver_orders) AS trusted_silver_orders,
    (SELECT SUM(daily_order_count)
     FROM gold_daily_restaurant_orders) AS gold_orders,
    (SELECT COUNT(*) FROM trusted_silver_orders)
      - (SELECT SUM(daily_order_count)
         FROM gold_daily_restaurant_orders) AS difference;
```

```python
%sql

SELECT
    SUM(CASE WHEN restaurant_id IS NULL THEN 1 ELSE 0 END) AS null_restaurant_id,
    SUM(CASE WHEN order_date IS NULL THEN 1 ELSE 0 END) AS null_order_date,
    SUM(CASE WHEN daily_order_count IS NULL THEN 1 ELSE 0 END) AS null_order_count,
    SUM(CASE WHEN daily_order_count <= 0 THEN 1 ELSE 0 END) AS invalid_order_count
FROM gold_daily_restaurant_orders;
```

```python
%sql

SELECT
    MIN(order_date) AS first_order_date,
    MAX(order_date) AS last_order_date,
    COUNT(DISTINCT order_date) AS number_of_days,
    COUNT(DISTINCT restaurant_id) AS number_of_restaurants,
    SUM(daily_order_count) AS total_orders
FROM gold_daily_restaurant_orders;
```

```python
%sql

SELECT
    order_date,
    SUM(daily_order_count) AS total_orders
FROM gold_daily_restaurant_orders
GROUP BY order_date
ORDER BY order_date;
```

```python
%sql
CREATE OR REPLACE TABLE gold_daily_order_trend AS
SELECT
    order_date,
    SUM(daily_order_count) AS total_orders
FROM gold_daily_restaurant_orders
GROUP BY order_date
ORDER BY order_date;
```

```python
%sql
SELECT *
FROM gold_daily_order_trend
ORDER BY order_date;
```
